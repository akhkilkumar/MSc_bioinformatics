```r
# Create a Numeric Vector
x <- c(10, 20, 30, 40, 50)
x
#Explanation: Creates a numeric vector using the c() function.
```

```r
#Create a Character Vector
fruits <- c("Apple", "Banana", "Mango", "Orange")
fruits
# Explanation: Creates a vector of character strings.
```

```r
#Create a Logical Vector
status <- c(TRUE, FALSE, TRUE, FALSE)
status
#Explanation: Creates a vector containing logical values.
```

```r
#. Find the Length of a Vector
x <- c(5, 10, 15, 20, 25)
length(x)

#Explanation: Returns the number of elements in the vector.
```

```r
# Access Vector Elements
x <- c(5, 10, 15, 20, 25)
x[1]
x[5]

#Explanation: Accesses the first and fifth elements of the vector.
```

```r
# Create a Sequence from 1 to 10
seq(1, 10)
#Explanation: Generates numbers from 1 to 10.
```

```r
# Create a Sequence with Step Size
seq(5, 50, by = 5)

#Explanation: Generates numbers from 5 to 50 with an interval of 5.
```

```r
# Create a Descending Sequence
seq(20, 1, by = -1)

#Explanation: Generates numbers from 20 down to 1.
```

```r
# Create 10 Equally Spaced Numbers

seq(0, 1, length.out = 10)

#Explanation: Generates 10 equally spaced numbers between 0 and 1.
```

```r
# Repeat a Number
rep(5, times = 10)
```

```r
# Repeat a Vector

rep(c(1, 2, 3), times = 4)

#Explanation: Repeats the vector four times.
```

```r
# Repeat Each Element

rep(c(10, 20, 30), each = 3)

#Explanation: Repeats each element three times before moving to the next.
```

```r
# Repeat Characters

rep(c("A", "B", "C"), times = 5)

#Explanation: Repeats the character vector five times.
```

```r
# Create a Mixed Pattern
rep(1:5, each = 2)
#Explanation: Repeats every number twice.
```

```r
# Create a Simple List
student <- list(Name = "Rahul", Age = 22, Marks = 89)
student

# Explanation: Creates a list containing different data types.
```

```r
#  Access a List Element
student$Name
student$age
student$Age

```

```r
# Create a List with Different Objects
mylist <- list(c(1,2,3), c("A","B"), c(TRUE,FALSE))
mylist
print('########')
mylist[3]

# Explanation: Creates a list containing numeric, character, and logical vectors.
```

```r
# Create a 2×3 Array
arr <- array(1:6, dim = c(2,3))
arr
```

```r
# Create a 3-D Array
arr3d <- array(1:36, dim = c(3,3,4))
print (arr3d)
```

```r
# create two 2 by 3 matrix
array1 <- array(c(1:12), dim = c(2,3,2))

print(array1)
#In the above example, we have used the array() function to create an array named array1. 
#Notice the arguments passed inside array(),
```

```r
print(array(c(1:15), dim = c(2,3,2)))
#Finally, the numbers from 1 to 12 that are arranged in two 2 by 3 matrices are printed.
```
