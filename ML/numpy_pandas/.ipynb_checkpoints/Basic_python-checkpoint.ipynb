```python
# Arithmetic Operations

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Modulus =", a % b)
print("Power =", a ** b)
```

```python
# Exercise: Calculate Area of a Rectangle
length = float(input("Enter length: "))
width = float(input("Enter width: "))

area = length * width

print("Area =", area)
```

```python
#Exercise: Swap Two Numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Before Swapping")
print(a)
print(b)

temp = a
a = b
b = temp

print("After Swapping")
print(a)
print(b)
```

```python
# Count Vowels
word = input("Enter word: ")

count = 0

for letter in word:
    if letter.lower() in "aeiou":
        count += 1

print("Vowels =", count)
```

```python
#Maximum Number
def maximum(a,b):
    if a>b:
        return a
    return b

print(maximum(20,35))
```

```python
# ATM PIN Verification (while)
pin = "1234"

while True:
    user_pin = input("Enter ATM PIN: ")

    if user_pin == pin:
        print("Login Successful")
        break
    else:
        print("Incorrect PIN")
```

```python
# Guess the Secret Number
secret = 15

while True:
    guess = int(input("Guess the number: "))

    if guess == secret:
        print("Correct!")
        break

    elif guess < secret:
        print("Too Small")

    else:
        print("Too Large")
```

```python
# Mobile Battery Simulation
import time

battery = 100

while battery > 0:
    time.sleep(3)
    print("Battery =", battery,"%")
    battery -= 10
```

```python
# Movie Ticket Booking
seats = 10

while seats > 0:

    input("Press Enter to Book Seat")

    seats -= 1

    print("Seats Left =", seats)
```

```python
seats = 100
while seats > 0:

    x=int(input("Press Enter to Book Seat"))
    
    seats=seats-x
    print(seats)
```

```python
# Password Generator
import random

letters = "abcdefghijklmnopqrstuvwxyz"

password = ""

for i in range(8):

    password += random.choice(letters)

print(password)
```
